## Part 1: Prompting Fundamentals

### 1.1 Zero-Shot Prompting

```python
# Example prompts for digital twin
zero_shot_examples = [
    {
        'task': 'activity_prediction',
        'prompt': 'Predict the next activity: I just finished work, had dinner, and now...',
        'expected': 'leisure/relaxation activity'
    },
    {
        'task': 'preference_extraction',
        'prompt': 'Extract preferences from: I love outdoor activities but hate crowded places',
        'expected': {'likes': ['outdoor', 'quiet'], 'dislikes': ['crowds']}
    }
]

print('Zero-Shot Examples:')
for ex in zero_shot_examples:
    print(f"\nTask: {ex['task']}")
    print(f"Prompt: {ex['prompt']}")
    print(f"Expected: {ex['expected']}")
```

### 🎯 Exercise 1.1 (Easy): Create Your Prompts

**Task**: Design prompts for your digital twin use case.

```python
my_prompts = [
    {
        'task': 'activity_prediction',
        'prompt': (
            "You are a digital twin of a person who works Monday to Friday, "
            "9am to 5pm, and exercises on some mornings. "
            "Predict their most likely next activity. "
            "Choose exactly one from: Sleep, Work, Exercise, Social, Leisure. "
            "Context: It is Tuesday, 6:30pm, and they just finished work. "
            "Answer with the activity name only."
        ),
        'expected': 'Leisure'
    },
    {
        'task': 'preference_extraction',
        'prompt': (
            "Extract preferences from the text below. "
            "Return only valid JSON with the keys 'likes' and 'dislikes', "
            "each a list of short strings.\n\n"
            "Text: I enjoy cooking at home and long walks, "
            "but I can't stand noisy restaurants."
        ),
        'expected': {'likes': ['cooking at home', 'walking'],
                     'dislikes': ['noisy restaurants']}
    },
    {
        'task': 'mood_classification',
        'prompt': (
            "Classify the mood of this message as exactly one of: "
            "positive, neutral, negative. "
            "Answer with one word only.\n\n"
            "Message: I got three hours of sleep and my deadline moved up."
        ),
        'expected': 'negative'
    },
    {
        'task': 'schedule_suggestion',
        'prompt': (
            "You are a digital twin that knows its owner prefers quiet "
            "evenings and exercises best in the morning. "
            "Suggest one activity for a free Saturday afternoon. "
            "Reply in one sentence and explain why it fits."
        ),
        'expected': 'A quiet activity such as reading or a walk, with a reason'
    },
]

for p in my_prompts:
    print(f"Task: {p['task']}")
    print(f"Prompt: {p['prompt']}")
    print(f"Expected: {p['expected']}\n")
```
